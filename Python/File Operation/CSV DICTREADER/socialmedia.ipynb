```python
# 1. Find the platform with the highest total views.

platform_summary = {}

for p in posts:
    platform=p.get("Platform")
    views =int(p["Views"])
    if platform in platform_summary:
        platform_summary[platform] += views

    else:
        platform_summary[platform]=views

platform_summary

max(platform_summary, key=lambda p: platform_summary[p])
```

```python

# Q2. Find the platform with the highest total likes

platform_summary = {}

for p in posts:
    platform = p.get("Platform")
    likes = int(p.get("Likes"))

    if platform in platform_summary:
        platform_summary[platform] += likes
    else:
        platform_summary[platform] = likes

print(platform_summary)
print(max(platform_summary, key=lambda p: platform_summary[p]))
```

```python

# Q3. Find the platform with the highest total shares

platform_summary = {}

for p in posts:
    platform = p.get("Platform")
    shares = int(p.get("Shares"))

    if platform in platform_summary:
        platform_summary[platform] += shares
    else:
        platform_summary[platform] = shares

print(platform_summary)
print(max(platform_summary, key=lambda p: platform_summary[p]))

```

```python
# Q4. Find the region with the highest total views

region_summary = {}

for p in posts:
    region = p.get("Region")
    views = int(p.get("Views"))

    if region in region_summary:
        region_summary[region] += views
    else:
        region_summary[region] = views

print(region_summary)
print(max(region_summary, key=lambda r: region_summary[r]))

```

```python
# Q16. Find posts where likes are greater than shares

result = []

for p in posts:
    likes = int(p.get("Likes"))
    shares = int(p.get("Shares"))

    if likes > shares:
        result.append(p)

print(result)
```

```python
# Q17. Find posts where Likes + Shares + Comments > 500,000

result = []

for p in posts:
    score = (
        int(p.get("Likes"))
        + int(p.get("Shares"))
        + int(p.get("Comments"))
    )

    if score > 500000:
        result.append(p)

print(result)
```

```python
# Q18. Platform having highest average views

d = {}

for p in posts:
    platform = p.get("Platform")

    if platform not in d:
        d[platform] = []

    d[platform].append(int(p.get("Views")))

avg = {}

for platform in d:
    avg[platform] = sum(d[platform]) / len(d[platform])

print(avg)
print(max(avg, key=lambda x: avg[x]))
```
