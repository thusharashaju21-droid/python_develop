```python
fr = open("movies.csv", "r", encoding="UTF-8")

movies = []

for line in fr:
    line = line.rstrip("\n")

    if line:
        id, title, language, year, run_time, rating, genre = line.split(",")

        dictionary = {
            "id": id,
            "title": title,
            "language": language,
            "year": year,
            "run_time": run_time,
            "rating": rating,
            "genre": genre
        }

        movies.append(dictionary)


print(len(movies))
```
