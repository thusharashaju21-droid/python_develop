Movies Data

```python
#Dataset Structure:
#[Title (str), Language (str), Duration in mins (int), Release Year (int), Rating (float)]
movies_data = [
     ["dangal", "hindi", 161, 2016, 8.3],
     ["premam", "malayalam", 156, 2015, 8.3],
     ["inception", "english", 148, 2010, 8.8],
     ["vikram", "tamil", 175, 2022, 8.3], 
     ["parasite", "korean", 132, 2019, 8.5], 
     ["rrr", "telugu", 187, 2022, 7.8], 
     ["drishyam", "malayalam", 160, 2013, 8.3],
     ["avatar", "english", 162, 2009, 7.9],
     ["kumbalangi nights", "malayalam", 135, 2019, 8.5],
     ["sholay", "hindi", 204, 1975, 8.1] 
]
```

```python
#Malayalam Movies
all_titiles=[m[0] for m in movies_data if m[1]=="malayalam"] 
all_titiles
```

```python
# all movies after year 2015
movies=[m[0] for m in movies_data if m[-2]>2015]
movies
```

```python
#Rating >=8.5 Count
rating_movie=[m[0] for m in movies_data if m[-2]>8.5]
rating_movie
```

```python
#Average Duration
duration=[m[2]for m in movies_data]
print(sum(duration))
```

```python
#Duration>150 and Rating>8
movie_dr=[m[0] for m in movies_data if m[2]>150 and m[-1]>8]
movie_dr
```

```python
#Highest Rated Movie
hightest=[m[0] for m in movies_data]
print(max(hightest))
```

```python
#oldest movie
lowest=[m[0] for m in movies_data]
print(min(lowest))
```
