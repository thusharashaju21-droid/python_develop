movie case study

```python
movies=[

    ["kgf","kannada",150,2005,8],
    ["balan","malayalam",130,2026,7],
    ["bangalore days","malayalam",150,2013,9],
    ["goatlife","malayalam",160,2024,9],
    ["abcd","malayalam",140,2008,6]
]
```

```python
#display all movie_titles
all_movie_titles= [m[0]for m in movies]
all_movie_titles
```

```python
#display all movie years
all_years={m[-2]for m in movies}
all_years
```

```python
#display all movie lanaguge
all_languages = [m[1] for m in movies]
all_languages
```

```python
#display all movies upper
all_movies={m[0].upper()for m in movies}
all_movies
```

```python
#movie titile starting with a
name= [m[0] for m in movies if m[0].startswith("a")]
name
```

```python
#display malayalam movies
malayalam_movies=[m[0] for m in movies if m[1]=="malayalam"]
malayalam_movies
```

```python
languages = [m[1] for m in movies]

lang_count = {l: languages.count(l) for l in languages}

lang_count
```

```python
import matplotlib.pyplot as plt

x=lang_count.keys()
y=lang_count.values()
plt.bar(x,y)

```

```python
years=[m[3] for m in movies]
year_count={y :years.count(y) for y in years}
year_count
```

```python
import matplotlib.pyplot as plt

x = year_count.keys()
y = year_count.values()

plt.bar(x, y)
```
