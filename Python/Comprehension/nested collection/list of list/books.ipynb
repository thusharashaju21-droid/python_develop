Books Data

```python
#Dataset Structure:
#[Title (str), Genre (str), Price in USD (int), Release Year (int), Rating (float)]
books_data = [
["The Hobbit","Fantasy",15,1937,4.8],
["Dune","Sci-Fi",18,1965,4.7],
["Project Hail Mary","Sci-Fi",22,2021,4.9],
["Atomic Habits","Self-Help",16,2018,4.8],
["1984","Dystopian",12,1949,4.6],
["The Silent Patient","Mystery",14,2019,4.5],
["Sapiens","History",20,2014,4.7],
["Harry Potter","Fantasy",25,1997,4.9],
["Educated","Memoir",17,2018,4.4],
["Klara and the Sun","Sci-Fi",13,2021,4.2]
]
```

```python
#Title
#all books more than 10 characters
title=[b[0] for b in books_data if len(b[0])>10 ]
title
```

```python
#all title start with"The"
title_start=[b[0] for b in books_data if b[0].startswith("The")]
title_start
```

```python
#Genre
# set-all unique  (without duplicate)
genres={b[1] for b in books_data}
genres
```

```python
# how many total books "SciFi " 
total_books=[b for b in books_data if b[1]=="Sci-Fi"]
total_books
```

```python
#Price
#total cost 
cost=[b[2] for b in books_data]
print(cost)
```

```python

# title and price cheapest book (lowest price)
low=[b[0] for b in books_data]
print(min(low))
```

```python

#Release Year
#using 2026 current year
years=[b[0] for b in books_data if 2026-b[3] ]
print(years)
```

```python
#all books befor year 2000
books=[b[0] for b in books_data if b[3]<2000]
print(books)
```

```python
#Rating
#average
ratings=[b[-1] for b in books_data]
print(sum(ratings))
```

```python
#count 4.8 or higher
high_count=[b for b in books_data if b[-1]>=4.8]
print(high_count)
```
