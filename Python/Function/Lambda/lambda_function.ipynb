```python
#lambda function for adding two number
add_number=lambda n1,n2 : n1+n2
print(add_number(100,200))
```

```python
#lamda function for substracting two number
sub_number=lambda n1,n2:n1-n2
print(sub_number(600,400))
```

```python
#lambda function for max_of_two number
max_two=lambda n1,n2:n1 if n1>n2 else n2
print(max_two(100,200))
```

```python
#lambda function for min_of_two number
min_two=lambda n1,n2:n1 if n1<n2 else n2
print(min_two(20,60))
```

```python
user_experice=[
["sooraj",1],
["kannan",2],
["thushara",6],
["kunja",5],
["sreya",3]
]
print(max(user_experice,key=lambda lst:lst[1]))
print(min(user_experice,key=lambda lst:lst[1]))
```
